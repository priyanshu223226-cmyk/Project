```python
from pathlib import Path
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

workspace = Path('..').resolve()
df = pd.read_csv(workspace / 'data' / 'raw' / 'Mall_Customers.csv')
df.head()
```

```python
print(df.shape)
df.info()
df.describe(include='all')
```

```python
gender_counts = df['Gender'].value_counts().sort_index()
gender_counts.plot(kind='bar', color=['#4C78A8', '#F58518'])
plt.title('Customer Count by Gender')
plt.xlabel('Gender')
plt.ylabel('Number of Customers')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
for gender in ['Female', 'Male']:
    subset = df[df['Gender'] == gender]
    plt.scatter(subset['Annual Income (k$)'], subset['Spending Score (1-100)'], label=gender, alpha=0.75)
plt.title('Annual Income versus Spending Score')
plt.xlabel('Annual Income (k$)')
plt.ylabel('Spending Score (1-100)')
plt.legend()
plt.tight_layout()
plt.show()
```
